```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import scipy
from scipy import stats

#this sets the size of the plot to something useful
plt.rcParams["figure.figsize"] = (15,10)
```

```python
#Creates a random normal distribution
#loc = center value for Guassian distribution
#scale = determines the width of the Guassian distribution: the larger the value the more spread out the distribution is
#size = number of variable generated

d = stats.norm.rvs(loc = 7.0, scale = .05, size = 10000000)
```

```python
#Plots a Guassian Distribution

fig, ax = plt.subplots(1, 1) #Creates the plot
ax.hist(d,50, density=True) #Plots a hitorgram (data, number of bins, makes it a density histogram)
plt.tick_params(labelsize = 24) #labels the axis
plt.xlim([6.7,7.3]) #sets a limit on the x-axis
x = np.linspace(6.8,7.2,1000) #returns 1000 evenly spaced numbers over a specified range (6.8-7.2)
ax.plot(x,stats.norm.pdf(x,loc = 7., scale = 0.05),linewidth = 8,alpha = 0.7) #creates a probability denstiy function at x centerd on the Guassian distribution and with the same width
plt.show() #plots the plot
```

```python
#Plots the Guassion on a one-sided log graph to show it makes a parabola
ax = plt.hist(d,50, density=True) #same histogram plot as above
plt.yscale('log') #makes the y-axis log scale
plt.tick_params(labelsize = 24) #changes font size of labels
plt.xlim([6.7,7.3]) #sets the x-axis limit
plt.show() #plots the plot
```

```python
#same plots as above, but adds a line to show the parabolic shape of the distrubtion
fig, ax = plt.subplots(1, 1)
ax.hist(d,50, density=True)
plt.yscale('log')
plt.tick_params(labelsize = 24)
plt.xlim([6.7,7.3])
x = np.linspace(6.7,7.3,1000)
ax.plot(x,stats.norm.pdf(x,loc = 7., scale = 0.05),linewidth = 8,alpha = 0.7)
plt.show()
```
